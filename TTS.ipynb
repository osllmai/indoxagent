```python
import sys
from PySide6.QtWidgets import (
    QApplication, QWidget, QPushButton, QVBoxLayout, QComboBox, QLabel
)
from PySide6.QtTextToSpeech import QTextToSpeech

class TTSWindow(QWidget):
    def __init__(self):
        super().__init__()
        self.setWindowTitle("Voice Selector")
        self.setGeometry(100, 100, 400, 150)

        self.tts = QTextToSpeech()
        self.voices = self.tts.availableVoices()

        self.text_to_speak = "Iran, officially the Islamic Republic of Iran, is a country in Western Asia, bordered by countries like Iraq, Turkey, Afghanistan, and Pakistan. Its capital is Tehran. Known for its rich history, Iran was once home to ancient civilizations like the Persian Empire. Persian (Farsi) is the official language, and Islam is the dominant religion. Iran is famous for its cultural heritage, including poetry, architecture, and art. It has vast natural resources, especially oil and gas. The country's diverse geography includes mountains, deserts, and the Caspian Sea coastline. Iran plays a significant role in regional politics and has a unique cultural identity."

        self.voice_selector = QComboBox()
        for voice in self.voices:
            self.voice_selector.addItem(voice.name())

        self.voice_selector.currentIndexChanged.connect(self.change_voice)

        self.speak_button = QPushButton("Speak")
        self.speak_button.clicked.connect(self.speak_text)

        layout = QVBoxLayout()
        layout.addWidget(QLabel("Select Voice:"))
        layout.addWidget(self.voice_selector)
        layout.addWidget(self.speak_button)
        self.setLayout(layout)

        # Set default voice
        if self.voices:
            self.tts.setVoice(self.voices[0])

    def change_voice(self, index):
        self.tts.setVoice(self.voices[index])

    def speak_text(self):
        self.tts.say(self.text_to_speak)

if __name__ == "__main__":
    app = QApplication(sys.argv)
    window = TTSWindow()
    window.show()
    sys.exit(app.exec())
```

```python
import sys
import PySide6
from PySide6.QtWidgets import (
    QApplication, QWidget, QVBoxLayout, QComboBox, QLabel, QPushButton
)
from PySide6.QtCore import qVersion, QTimer
from PySide6.QtTextToSpeech import QTextToSpeech

class SiriV2(QWidget):
    def __init__(self):
        super().__init__()
        self.setWindowTitle("Siri v2 - Voice Assistant")
        self.setGeometry(200, 200, 400, 230)

        # Initialize TTS
        self.tts = QTextToSpeech()
        self.tts.stateChanged.connect(self.handle_tts_state)

        # Log version info
        print("==== System Info ====")
        print("PySide6 version:", PySide6.__version__)
        print("Qt version:", qVersion())
        print("TTS Engine:", self.tts.engine())
        print("=====================")

        self.voices = self.tts.availableVoices()
        if not self.voices:
            print("No voices found!")
        else:
            print("Available Voices:")
            for voice in self.voices:
                print(f"- {voice.name()} ({voice.locale().name()})")

        self.response = (
            "Iran, officially the Islamic Republic of Iran, is a country in Western Asia. "
            "It has a rich history as the heart of the Persian Empire. Tehran is its capital. "
            "The country is known for its culture, poetry, architecture, and vast oil reserves."
        )

        self.voice_selector = QComboBox()
        for voice in self.voices:
            display_name = f"{voice.name()} ({voice.locale().name()})"
            self.voice_selector.addItem(display_name)
        self.voice_selector.currentIndexChanged.connect(self.set_voice)

        self.label = QLabel("Select a voice and control Siri")

        self.ask_button = QPushButton("Ask Siri")
        self.ask_button.clicked.connect(self.speak_response)

        self.pause_button = QPushButton("Pause Siri")
        self.pause_button.clicked.connect(self.pause_speech)
        self.pause_button.setEnabled(False)

        self.resume_button = QPushButton("Resume Siri")
        self.resume_button.clicked.connect(self.resume_speech)
        self.resume_button.setEnabled(False)

        self.stop_button = QPushButton("Stop Siri")
        self.stop_button.clicked.connect(self.stop_speech)
        self.stop_button.setEnabled(False)

        layout = QVBoxLayout()
        layout.addWidget(self.label)
        layout.addWidget(self.voice_selector)
        layout.addWidget(self.ask_button)
        layout.addWidget(self.pause_button)
        layout.addWidget(self.resume_button)
        layout.addWidget(self.stop_button)
        self.setLayout(layout)

        if self.voices:
            self.tts.setVoice(self.voices[0])

    def set_voice(self, index):
        self.tts.setVoice(self.voices[index])

    def speak_response(self):
        self.ask_button.setEnabled(False)
        self.stop_button.setEnabled(True)
        self.pause_button.setEnabled(True)
        self.resume_button.setEnabled(False)
        self.tts.say(self.response)

    def stop_speech(self):
        if self.tts.state() != QTextToSpeech.State.Ready:
            QTimer.singleShot(100, self.tts.stop)

    def pause_speech(self):
        if self.tts.state() == QTextToSpeech.State.Speaking:
            self.tts.pause()
            self.pause_button.setEnabled(False)
            self.resume_button.setEnabled(True)

    def resume_speech(self):
        if self.tts.state() == QTextToSpeech.State.Paused:
            self.tts.resume()
            self.pause_button.setEnabled(True)
            self.resume_button.setEnabled(False)

    def handle_tts_state(self, state):
        if state == QTextToSpeech.State.Ready:
            self.ask_button.setEnabled(True)
            self.stop_button.setEnabled(False)
            self.pause_button.setEnabled(False)
            self.resume_button.setEnabled(False)
        elif state == QTextToSpeech.State.Speaking:
            self.ask_button.setEnabled(False)
            self.stop_button.setEnabled(True)
            self.pause_button.setEnabled(True)
            self.resume_button.setEnabled(False)
        elif state == QTextToSpeech.State.Paused:
            self.ask_button.setEnabled(False)
            self.stop_button.setEnabled(True)
            self.pause_button.setEnabled(False)
            self.resume_button.setEnabled(True)

if __name__ == "__main__":
    app = QApplication(sys.argv)
    window = SiriV2()
    window.show()
    sys.exit(app.exec())
```
